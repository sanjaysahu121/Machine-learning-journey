# Polynomial Regression 

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# generated same syntheic datapoint or reproducibility of result 
np.random.seed(1)
```

```python
# generate syntheic X 
x = 2*np.random.rand(100,1)
x
```

```python
# generate random synthic y using x
y = 4+3*x+1.5*x**2+np.random.randn(100,1)
y
```

```python
# train test split 


from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test=train_test_split(x,y , test_size = 0.2 , random_state = 1)
```

```python
x_train.shape , x_test.shape, y_train.shape , y_test.shape
```

```python
# perform polynomial regression 

from sklearn.preprocessing import PolynomialFeatures
```

```python
poly_features = PolynomialFeatures(degree = 2 , include_bias = False)
poly_features
```

```python
x_ploy_train = poly_features.fit_transform(x_train)
x_ploy_train
```

```python
from sklearn.linear_model import LinearRegression
ploy_reg = LinearRegression()
ploy_reg
```

```python
ploy_reg.fit(x_ploy_train , y_train)
```

```python
ploy_reg.coef_
```

```python
ploy_reg.intercept_
```

```python
# make prediction on train data itself for evalution metrics

y_poly_preddict = ploy_reg.predict(x_ploy_train)
y_poly_preddict
```

```python
from sklearn.metrics import mean_squared_error
mean_squared_error(y_train , y_poly_preddict)
```

```python
plt.scatter(x_train , y_train , label  = 'training data')
```

```python
degree = 2
x_range = np.linspace(0,2,100).reshape(-1,1)
x_range_poly = poly_features.transform(x_range)
plt.plot(x_range, ploy_reg.predict(x_range_poly),color = 'red', label = f'Polynomial Regression (Degree{degree})')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Polynomial Regresion')
plt.legend()
plt.show()
```
